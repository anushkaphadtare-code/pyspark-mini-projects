we have 3 datasets:

2. orders: Order ID, customer ID, product ID, date, quantity, unit price

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType,
    DoubleType
)
```

```python
# define schema of orders dataset:
orders_schema= StructType([
    StructField("order_id", StringType(), False),
    StructField("customer_id", StringType(), False),
    StructField("product_id", StringType(), False),
    StructField("order_date",StringType(), True),
    StructField("quantity", IntegerType(),True),
    StructField("unit_price", DoubleType(), True)
])

# prepare sample records:
orders_data = [
    ("O101", "C001", "P001", "2026-01-05", 2, 500.0),
    ("O102", "C002", "P002", "2026-01-08", 1, 1200.0),
    ("O103", "C001", "P003", "2026-01-15", 3, 250.0),
    ("O104", "C003", "P001", "2026-02-02", 1, 500.0),
    ("O105", "C004", "P004", "2026-02-10", 2, 800.0),
    ("O106", "C002", "P003", "2026-02-18", 4, 250.0),
    ("O107", "C005", "P002", "2026-03-03", 2, 1200.0),
    ("O108", "C003", "P004", "2026-03-12", 1, 800.0),
    ("O109", "C004", "P001", "2026-03-20", 3, 500.0),
    ("O110", "C005", "P003", "2026-03-25", 2, 250.0)
]

# create spark dataframe:
orders_df = spark.createDataFrame(
    orders_data,
    schema = orders_schema
)

# display the dataframe:
display(orders_df)
```

```python
# number of records:
print("Total number of records:", orders_df.count())

# column names:
print("Column names: ",orders_df.columns)

#printSchema():
orders_df.printSchema()

# first five records:
orders_df.show(5, truncate = False)
```

```python

from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    DoubleType
)

# Define the schema
products_schema = StructType([
    StructField("product_id", StringType(), False),
    StructField("product_name", StringType(), False),
    StructField("category", StringType(), True),
    StructField("unit_price", DoubleType(), True)
])

# Prepare sample records
products_data = [
    ("P001", "Wireless Mouse", "Electronics", 500.0),
    ("P002", "Keyboard", "Electronics", 1200.0),
    ("P003", "Notebook", "Stationery", 250.0),
    ("P004", "Backpack", "Accessories", 800.0)
]

# Create the Spark DataFrame
products_df = spark.createDataFrame(
    products_data,
    schema=products_schema
)

# Display the DataFrame
display(products_df)
```

```python

print("Total products:", products_df.count())

print("Columns:", products_df.columns)

products_df.printSchema()

products_df.show(truncate=False)
```

```python

from pyspark.sql.types import (
    StructType,
    StructField,
    StringType
)

# Define the schema
customers_schema = StructType([
    StructField("customer_id", StringType(), False),
    StructField("customer_name", StringType(), False),
    StructField("region", StringType(), True)
])

# Prepare sample records
customers_data = [
    ("C001", "Aarav Sharma", "West"),
    ("C002", "Priya Patel", "South"),
    ("C003", "Rohan Mehta", "North"),
    ("C004", "Ananya Desai", "East"),
    ("C005", "Ishaan Kulkarni", "West")
]

# Create the Spark DataFrame
customers_df = spark.createDataFrame(
    customers_data,
    schema=customers_schema
)

# Display the DataFrame
display(customers_df)
```

```python

print("Total customers:", customers_df.count())

print("Columns:", customers_df.columns)

customers_df.printSchema()

customers_df.show(truncate=False)
```

```python

print("Orders:", orders_df.count())
print("Products:", products_df.count())
print("Customers:", customers_df.count())
```

```python
# check for missing values(NULLs):

from pyspark.sql.functions import col, sum, when

null_counts = orders_df.select(
    [
        sum(
            when(col(c).isNull(), 1).otherwise(0)
        ).alias(c)
        for c in orders_df.columns
    ]
)

display(null_counts)
```

```python
# check for duplicate records:

total_rows = orders_df.count()

distinct_rows = orders_df.distinct().count()

duplicate_rows = total_rows - distinct_rows

print("Total rows:", total_rows)
print("Distinct rows:", distinct_rows)
print("Duplicate rows:", duplicate_rows)
```

```python

from pyspark.sql.functions import col, sum, when

product_null_counts = products_df.select(
    [
        sum(
            when(col(c).isNull(), 1).otherwise(0)
        ).alias(c)
        for c in products_df.columns
    ]
)

display(product_null_counts)
```

```python

# Check for completely identical rows
total_products = products_df.count()
distinct_products = products_df.distinct().count()

print("Total rows:", total_products)
print("Distinct rows:", distinct_products)
print("Excess duplicate rows:", total_products - distinct_products)

# Check for duplicate product IDs
duplicate_product_ids = (
    products_df
    .groupBy("product_id")
    .count()
    .filter(col("count") > 1)
)

display(duplicate_product_ids)
```

```python

customer_null_counts = customers_df.select(
    [
        sum(
            when(col(c).isNull(), 1).otherwise(0)
        ).alias(c)
        for c in customers_df.columns
    ]
)

display(customer_null_counts)
```

```python

# Check completely identical rows
total_customers = customers_df.count()
distinct_customers = customers_df.distinct().count()

print("Total rows:", total_customers)
print("Distinct rows:", distinct_customers)
print("Excess duplicate rows:", total_customers - distinct_customers)

# Check duplicate customer IDs
duplicate_customer_ids = (
    customers_df
    .groupBy("customer_id")
    .count()
    .filter(col("count") > 1)
)

display(duplicate_customer_ids)
```

```python

from pyspark.sql.functions import col

orders_transformed_df = orders_df.withColumn(
    "total_amount",
    col("quantity") * col("unit_price")
)

display(orders_transformed_df)
```

```python

orders_transformed_df.select(
    "order_id",
    "quantity",
    "unit_price",
    "total_amount"
).show()
```

```python

from pyspark.sql.functions import sum

product_revenue_df = (
    orders_transformed_df
    .groupBy("product_id")
    .agg(
        sum("total_amount").alias("total_revenue")
    )
)

display(product_revenue_df)
```

```python

from pyspark.sql.functions import sum, count

product_summary_df = (
    orders_transformed_df
    .groupBy("product_id")
    .agg(
        sum("total_amount").alias("total_revenue"),
        count("order_id").alias("number_of_orders")
    )
)

display(product_summary_df)
```

```python

product_revenue_details_df = product_summary_df.join(
    products_df,
    on="product_id",
    how="inner"
)

display(product_revenue_details_df)
```

```python

from pyspark.sql.functions import col

product_revenue_ranked_df = product_revenue_details_df.orderBy(
    col("total_revenue").desc()
)

display(product_revenue_ranked_df)
```

```python

orders_customers_df = orders_transformed_df.join(
    customers_df,
    on="customer_id",
    how="inner"
)

display(orders_customers_df)
```

```python

from pyspark.sql.functions import sum, count, col

region_revenue_df = (
    orders_customers_df
    .groupBy("region")
    .agg(
        sum("total_amount").alias("total_revenue"),
        count("order_id").alias("number_of_orders")
    )
    .orderBy(col("total_revenue").desc())
)

display(region_revenue_df)
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS retail_sales;
```

```python

catalog_name = "workspace"
schema_name = "retail_sales"

silver_table = f"{catalog_name}.{schema_name}.retail_orders_silver"

(
    orders_transformed_df
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(silver_table)
)

print(f"Saved table: {silver_table}")
```

```python

product_gold_table = (
    f"{catalog_name}.{schema_name}.retail_product_revenue_gold"
)

region_gold_table = (
    f"{catalog_name}.{schema_name}.retail_region_revenue_gold"
)

(
    product_revenue_ranked_df
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(product_gold_table)
)

(
    region_revenue_df
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(region_gold_table)
)

print("Saved all Gold tables successfully.")
```
